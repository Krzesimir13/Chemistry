##Uploading the file

```python
# Function to open and convert file to df
# folder is the folder with exp files
# name is the filename
def text_to_df(folder, name):
  parent = Path('/content/drive/MyDrive/Chemistry')
  file = parent / folder / name
  df = pd.read_csv(
      file,
      sep=r'\s+',
      skiprows=1,  # <-- fixes the header problem
      names=["Index", "Field_G", "Abs"]
      )

  # Convert to numeric (just in case)
  df["Field_G"] = pd.to_numeric(df["Field_G"], errors='coerce')
  df["Abs"] = pd.to_numeric(df["Abs"], errors='coerce')

  # Drop any rows that contain NaN after conversion
  df = df.dropna()

  return df
```
