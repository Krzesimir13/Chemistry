```python
#encoding of data to DataFrame
from pathlib import Path


def text_to_df(name):
  parent = Path('/content/drive/MyDrive/Chemistry/KK154 EPR')
  file = parent / name
  df = pd.read_csv(
      file,
      sep=r'\s+',
      skiprows=1,  # <-- fixes the header problem
      names=["Index", "Field_G", "Abs"]
      )

  # Convert to numeric (just in case)
  df["Field_G"] = pd.to_numeric(df["Field_G"], errors='coerce')
  df["Abs"] = pd.to_numeric(df["Abs"], errors='coerce')

  # Drop any rows that contain NaN after conversion
  df = df.dropna()

  return df
```

```python
file_1 = 'KK154-8G-13dB-4scans-1.txt'
exp_1 = text_to_df(file_1)
file_2 = 'KK154-8G-13dB-4scans-130K-ZOOM.txt'
exp_2 = text_to_df(file_2)
file_3 = 'KK154-G-13dB-4scans-130K-ZOOM.txt'
exp_3 = text_to_df(file_3)
#print(exp_1.head())
```

```python
# Plot
import plotly.express as px


def make_plot(df):
  fig = px.line(df,
                x="Field_G",
                y="Abs",
                title="KK154",
                labels={"Field_G": "Field [G]", "Abs": "Absorption [a.u.]"}
              )
  fig.show()
```

The plot from exp_1 looks the best.

```python
make_plot(exp_1)
```

Code for a plot wit g-coefficient in x-axis

```python
# --- Convert field → g using extracted frequency ---
mw_freq = 9.419598  # GHz
exp_1["g"] = 714.48 * mw_freq / exp_1["Field_G"]
exp_1.head()
```

```python
def make_plot_g(df):
    fig = px.line(df,
                  x="g",
                  y="Abs",
                  title="KK154",
                  labels={"g": "g-value", "Abs": "Absorption [a.u.]"}
                 )
    # Disable autorange to allow manual zoom
    fig.update_xaxes(autorange=False)

    # Zoom around g = 2
    fig.update_xaxes(range=[1.5, 2.5])

    fig.show()

make_plot_g(exp_1)
```
